## Project Overview

This project analyzes 14 days of sales and inventory
data for Wanjiku's fresh-produce stall at Gikomba Market.

The objective is to evaluate business performance,
identify sales trends, assess stock levels, calculate
product profitability, and provide data-driven
recommendations for improving the business.

**Tools:** Python and Google Colab.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2


# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

```python
print("Number of sales records:", len(daily_sales))
print("Number of day labels:", len(days))
print("Number of products:", len(stock))
```

## Task 1: Overall Sales Performance

In this section, I will calculate the total and average
daily sales over the 14-day period. I will also identify
the highest and lowest sales amounts and determine the
corresponding day names and weeks.

```python
# Calculate overall sales statistics

total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)

highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

# Find the positions of the highest and lowest sales
best_day_index = daily_sales.index(highest_sales)
worst_day_index = daily_sales.index(lowest_sales)

# Identify the day names and week numbers
best_day = days[best_day_index]
worst_day = days[worst_day_index]

best_week = (best_day_index // 7) + 1
worst_week = (worst_day_index // 7) + 1

# Display the results
print("OVERALL SALES PERFORMANCE")
print("-" * 35)
print(f"Total sales: KES {total_sales:,.2f}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest sales: KES {highest_sales:,.2f}")
print(f"Lowest sales: KES {lowest_sales:,.2f}")
print(f"Best sales day: {best_day}, Week {best_week}")
print(f"Worst sales day: {worst_day}, Week {worst_week}")
```

## Task 2: Weekly Sales and Growth

In this section, I will calculate the total sales for
each week using a reusable function. I will then compare
the two weeks to determine the percentage increase or
decrease in sales.

```python
# Calculate sales totals for each week

def week_totals(sales):
    """
    Calculate and return the sales totals for Week 1 and Week 2.

    Args:
        sales: A list containing 14 daily sales values.

    Returns:
        A tuple containing Week 1 and Week 2 sales totals.
    """
    week1 = sum(sales[:7])
    week2 = sum(sales[7:14])

    return week1, week2


# Call the function and unpack the returned values
week1_total, week2_total = week_totals(daily_sales)

# Calculate week-on-week growth percentage
growth_percentage = ((week2_total - week1_total) / week1_total) * 100

# Display results
print("WEEKLY SALES ANALYSIS")
print("-" * 35)
print(f"Week 1 sales: KES {week1_total:,.2f}")
print(f"Week 2 sales: KES {week2_total:,.2f}")
print(f"Week-on-week growth: {growth_percentage:.2f}%")
```

## Task 3: Sales Day Classification

In this section, I will classify each sales day as Strong,
Normal or Weak based on its performance against the
average daily sales.

A Strong day records sales more than 20% above average,
while a Weak day records sales more than 20% below
average. All other days are classified as Normal.

```python
def classify_day(amount, average):
    """
    Classify daily sales based on the average sales amount.

    Strong: More than 20% above average.
    Weak: More than 20% below average.
    Normal: Within the 20% thresholds.
    """
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"


# Create a dictionary to count each category
day_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

# Classify every day's sales
for day, amount in zip(days, daily_sales):
    category = classify_day(amount, average_sales)
    day_counts[category] += 1

    print(f"{day}: KES {amount:,.2f} - {category}")

# Display category totals
print("\nSALES CATEGORY SUMMARY")
print("-" * 35)

for category, count in day_counts.items():
    print(f"{category} days: {count}")
```

## Task 4: Weekday vs Weekend Performance

In this section, I will compare average sales during
weekdays (Monday to Friday) and weekends (Saturday
and Sunday). I will calculate the percentage difference
to determine which period generates higher sales.

```python
# Separate weekday and weekend sales

weekday_sales = []
weekend_sales = []

for day, amount in zip(days, daily_sales):
    if day in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
        weekday_sales.append(amount)
    else:
        weekend_sales.append(amount)

# Calculate average sales
weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

# Calculate percentage difference
percentage_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

# Display results
print("WEEKDAY VS WEEKEND PERFORMANCE")
print("-" * 40)
print(f"Average weekday sales: KES {weekday_average:,.2f}")
print(f"Average weekend sales: KES {weekend_average:,.2f}")
print(f"Percentage difference: {percentage_difference:.2f}%")

if weekend_average > weekday_average:
    print("Weekend sales perform better.")
else:
    print("Weekday sales perform better.")
```

## Task 5: Stock Management and Gross Margins

In this section, I will analyze the available stock for
each product against its reorder level. I will create a
reusable function to identify products that need
restocking and calculate the gross margin percentage
for every product.

```python
# Identify products that need restocking

def reorder_products(stock):
    """
    Return products whose stock quantity is at or below
    their specified reorder level.

    Args:
        stock: Dictionary containing product stock details.

    Returns:
        A list of product names requiring restocking.
    """
    products_to_reorder = []

    for product, details in stock.items():
        if details["qty"] <= details["reorder"]:
            products_to_reorder.append(product)

    return products_to_reorder


# Identify products requiring restocking
reorder_list = reorder_products(stock)

# Calculate gross margin for each product
print("STOCK AND GROSS MARGIN ANALYSIS")
print("-" * 55)

for product, details in stock.items():
    gross_margin = (
        (details["price"] - details["cost"]) / details["price"]
    ) * 100

    print(f"\nProduct: {product}")
    print(f"Quantity in stock: {details['qty']} kg")
    print(f"Reorder level: {details['reorder']} kg")
    print(f"Gross margin: {gross_margin:.2f}%")

print("\nPRODUCTS REQUIRING RESTOCKING")
print("-" * 35)

for product in reorder_list:
    print(product)
```

## Task 6: Stall Performance Report

In this section, I will combine the results from the
previous tasks into one formatted business report.
The report will summarize sales performance, weekly
growth, daily classifications, weekday and weekend
performance, stock levels, and product gross margins.

```python
# STALL PERFORMANCE REPORT

print("=" * 55)
print("             WANJIKU'S STALL PERFORMANCE REPORT")
print("=" * 55)

# Overall sales performance
print("\n1. OVERALL SALES PERFORMANCE")
print("-" * 55)
print(f"Total sales:             KES {total_sales:,.2f}")
print(f"Average daily sales:     KES {average_sales:,.2f}")
print(f"Highest daily sales:     KES {highest_sales:,.2f}")
print(f"Lowest daily sales:      KES {lowest_sales:,.2f}")
print(f"Best sales day:          {best_day}, Week {best_week}")
print(f"Worst sales day:         {worst_day}, Week {worst_week}")

# Weekly performance
print("\n2. WEEKLY PERFORMANCE")
print("-" * 55)
print(f"Week 1 sales:            KES {week1_total:,.2f}")
print(f"Week 2 sales:            KES {week2_total:,.2f}")
print(f"Weekly growth:           {growth_percentage:.2f}%")

# Daily classification
print("\n3. DAILY SALES CLASSIFICATION")
print("-" * 55)
for category, count in day_counts.items():
    print(f"{category} days:             {count}")

# Weekday vs weekend
print("\n4. WEEKDAY VS WEEKEND PERFORMANCE")
print("-" * 55)
print(f"Average weekday sales:   KES {weekday_average:,.2f}")
print(f"Average weekend sales:   KES {weekend_average:,.2f}")
print(f"Weekend advantage:       {percentage_difference:.2f}%")

# Stock and gross margins
print("\n5. STOCK AND GROSS MARGIN ANALYSIS")
print("-" * 55)

for product, details in stock.items():
    margin = (
        (details["price"] - details["cost"]) / details["price"]
    ) * 100

    status = "REORDER" if product in reorder_list else "Sufficient"

    print(
        f"{product:<12} | "
        f"Stock: {details['qty']:>2} kg | "
        f"Margin: {margin:>5.2f}% | "
        f"{status}"
    )

print("\n6. PRODUCTS REQUIRING RESTOCKING")
print("-" * 55)

for product in reorder_list:
    print(f"- {product}")

print("\n" + "=" * 55)
print("             END OF PERFORMANCE REPORT")
print("=" * 55)
```
